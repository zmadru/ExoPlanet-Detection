```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
classes = {
    0: 'False Positive',
    1: 'Confirmed Planet',
}
candidates = pd.read_csv('candidates_predictions_with_coords.csv')
candidates.head()
```

```python
all_data = pd.read_csv('../csv/cumulative_2025.11.25_08.30.17.csv', skiprows=144)
all_data.head()
```

```python
# merge candidates with all_data to get the star names and magnitudes
magnitudes_cols = [col for col in all_data.columns if 'mag' in col]
magnitudes_cols

candidates = candidates.merge(all_data[['kepoi_name'] + magnitudes_cols], on='kepoi_name' , how='left', validate='one_to_one')
candidates.head()
```

```python
candidates['predicted'] = candidates['predicted'].map(classes)
candidates.head()
```

```python
candidates.to_csv('candidates_predictions_with_coords_and_magnitudes.csv', index=False)
```

```python
import pandas as pd
df_2022 = pd.read_csv('../csv/cumulative_2022.09.30_09.06.43.csv', skiprows=144)
df_2025 = pd.read_csv('../csv/cumulative_2025.11.25_08.30.17.csv', skiprows=144)

candidates_2022 = df_2022[df_2022['koi_disposition'] == 'CANDIDATE']
candidates_2025 = df_2025[df_2025['koi_disposition'] == 'CANDIDATE']

print(candidates_2025.shape[0], candidates_2022.shape[0])
print(candidates_2022.shape[0] - candidates_2025.shape[0])
id_2022 = candidates_2022['kepoi_name'].unique()
id_2025 = candidates_2025['kepoi_name'].unique()
print(len(id_2022), len(id_2025))
print(len(id_2022) - len(id_2025))
changed_candidates = candidates_2022[~candidates_2022['kepoi_name'].isin(id_2025)]
changed_candidates.shape

changed_candidates.to_csv('../results/changed_candidates.csv', index=False)

```
